## 1. Lists: ordered and mutable

Lists preserve order and can change. Useful operations include `append`, `extend`, `insert`, `remove`, `pop`, indexing, slicing, `len`, `sum`, `min`, `max`, `sorted` and `reversed`.


### Worked example

Predict the result first. Then run the cell and change one input.


```python
amounts = [18.5, 42.0, 7.5]
amounts.append(12.0)
first_two = amounts[:2]
largest = max(amounts)
ascending = sorted(amounts)
print(amounts, first_two, largest, ascending)
```

### Guided exercise 1

```python
expenses = [25.0, 8.5, 41.5]
# TODO: append 15.0
expenses.append(15.0)
# TODO: remove 8.5
expenses.remove(8.5)
# TODO: last_two
last_two = expenses[-2:]
# TODO: count, total and average
count = len(expenses)
total = sum(expenses)
average = total / count
print(expenses)
print("Last two:", last_two)
print("Count:", count, "| Total:", total, "| Average:", round(average, 2))

# Checks
assert expenses == [25.0, 41.5, 15.0]
assert last_two == [41.5, 15.0]
assert count == 3 and total == 81.5
assert round(average, 2) == 27.17
```

## 2. Tuples: fixed records and unpacking

Tuples are ordered but immutable. Use them for fixed-position values and unpack them into meaningful names. `count()` and `index()` also work on tuples.


### Worked example

Predict the result first. Then run the cell and change one input.


```python
transaction = ("T001", "food", 18.5)
tx_id, category, amount = transaction
print(tx_id, category, amount)
print(transaction.index("food"))
```

### Guided exercise 2

Unpack a product tuple. Calculate its stock value. Then use starred unpacking on a tuple of monthly values.


```python
product = ("P01", "Keyboard", 49.99, 12)
# TODO: unpack sku, name, price, stock
sku, name, price, stock = product
# TODO: calculate stock_value
stock_value = price * stock
print(sku, name, price, stock)
print("Stock value:", round(stock_value, 2))

months = (120, 135, 128, 150)
# TODO: unpack first, *middle, last
first, *middle, last = months
print("First:", first, "| Middle:", middle, "| Last:", last)

# Checks
assert sku == "P01" and name == "Keyboard"
assert round(stock_value, 2) == 599.88
assert first == 120 and middle == [135, 128] and last == 150
```

## 3. Sets: uniqueness and membership

Sets store unique values. Use `add`, `discard`, union `|`, intersection `&`, difference `-`, `issubset()` and membership with `in`. Sets do not guarantee display order.


### Worked example

Predict the result first. Then run the cell and change one input.


```python
required = {"python", "sql", "git"}
team = {"python", "excel", "git", "python"}
missing = required - team
shared = required & team
all_skills = required | team
print(missing, shared, all_skills)
```

### Guided exercise 3

Compare required and observed customer categories. Find missing, unexpected and shared categories. Add one category safely.


```python
required = {"food","travel","software"}
observed = {"food","health","travel","food"}
# TODO: missing, unexpected, shared
missing = required - observed      # required but not observed
unexpected = observed - required   # observed but not required
shared = required & observed       # in both sets
print("Missing:", sorted(missing))
print("Unexpected:", sorted(unexpected))
print("Shared:", sorted(shared))
# TODO: add "software"
observed.add("software")           # add() is safe: no duplicates possible
print(observed)

# Checks
assert missing == {"software"}
assert unexpected == {"health"}
assert shared == {"food", "travel"}
assert required.issubset(observed)
assert len(observed) == 4
```

## 4. Dictionaries: labelled data

Dictionaries map unique keys to values. Use direct indexing for required keys, `get()` for optional keys, `keys()`, `values()`, `items()`, `update()`, `setdefault()` and `pop()`.


### Worked example

Predict the result first. Then run the cell and change one input.


```python
customer = {"id":"C01","name":"Ana","segment":"premium"}
print(customer["name"])
print(customer.get("country","unknown"))
customer.update({"country":"ES"})
for key,value in customer.items():
    print(key,value)
```

### Guided exercise 4

Create a product dictionary, update stock, add tags with `setdefault`, and remove an obsolete key with `pop`.


```python
product = {"id":"P1","name":"Monitor","price":199.0,"stock":4,"legacy_code":"OLD"}
# TODO: increase stock by 3
product["stock"] += 3
# TODO: create tags with setdefault and append "office"
product.setdefault("tags", []).append("office")
# TODO: remove legacy_code with pop
removed = product.pop("legacy_code")
print("Removed:", removed)
print(product)

# Checks
assert product["stock"] == 7
assert product["tags"] == ["office"]
assert "legacy_code" not in product
assert removed == "OLD"
```

## 5. Nested collections

Real data often combines collections: a list of dictionaries or a dictionary whose values are lists. Traverse one level at a time and keep every record shape consistent.


### Worked example

Predict the result first. Then run the cell and change one input.


```python
orders = [
    {"id":"O1","customer":"C1","items":[12.0,8.0]},
    {"id":"O2","customer":"C2","items":[20.0]},
]
for order in orders:
    order["total"] = sum(order["items"])
print(orders)
```

### Guided exercise 5

Build a dictionary indexed by contact ID from a list of contacts. Then find all IDs tagged as clients.


```python
contacts = [
    {"id":"C1","name":"Ana","tags":{"client","newsletter"}},
    {"id":"C2","name":"Luis","tags":{"supplier"}},
]
# TODO: by_id dictionary
by_id = {}
for contact in contacts:
    by_id[contact["id"]] = contact
# TODO: client_ids list
client_ids = []
for contact_id, contact in by_id.items():
    if "client" in contact["tags"]:
        client_ids.append(contact_id)
print(contacts)
print("By ID:", by_id)
print("Client IDs:", client_ids)

# Checks
assert set(by_id.keys()) == {"C1", "C2"}
assert by_id["C2"]["name"] == "Luis"
assert client_ids == ["C1"]
```

## Final guided challenge

Create a grade tracker: dictionary indexed by student ID, list of grades per student, a set of subjects, average calculation, and ranking that safely handles students with no grades.

Build it in small steps and keep every previous check passing.


```python
students = {
    "S1":{"name":"Ana","grades":[8,9],"subjects":{"python","data"}},
    "S2":{"name":"Luis","grades":[],"subjects":{"python"}},
}
# TODO: average(student_id) -> float or None
def average(student_id):
    grades = students[student_id]["grades"]
    if len(grades) == 0:
        return None            # safe: no division by zero
    return sum(grades) / len(grades)

# TODO: ranked list excluding students without grades
ranked = []
for student_id in students:
    avg = average(student_id)
    if avg is not None:
        ranked.append((student_id, students[student_id]["name"], avg))
ranked.sort(key=lambda row: row[2], reverse=True)   # highest average first

# TODO: unique subjects taught
subjects = set()
for student in students.values():
    subjects = subjects | student["subjects"]

for position, (student_id, name, avg) in enumerate(ranked, start=1):
    print(f"{position}. {name} ({student_id}): {avg:.2f}")
print("Unique subjects:", sorted(subjects))

# TODO: checks
assert average("S1") == 8.5
assert average("S2") is None
assert ranked == [("S1", "Ana", 8.5)]
assert subjects == {"python", "data"}
print("All checks passed")
```
